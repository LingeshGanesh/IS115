# **Optimization Problems**

An optimization problem is associated with an objective function. This means that:

- For any solution, the objective function evaluates a value
- The value indicates the goodness of the solution
- The goal is to obtain the solution with as optimal a value as possible.

### **Ideal Properties**

- Fast: Able to run in polynomial time complexity
- General: Works well on all instances of a problem
- Optimal: Finds the exact optimal solution

### **Trade-offs**
1) General + Optimal
    - very slow
    - e.g., brute force (exhaustive search)
2) Fast + Optimal
    - may not cover all cases
3) Fast + General
    - Not guaranteed to find optimal solution
    - Can we find a solution that is good enough?
    - Yes, with heuristics strategies


### **Heuristics Techniques**
### Greedy Algorithm
Greedy Algorithms are a class of algorithms that make locally optimal choices at each step with the hope of finding a global optimum solution. 



A greedy algorithm works by making a sequence of choices, each of which being the most optimal at the moment. It is important to note that greedy algorithms do not consider past decisions, instead selecting the locally optimal choice at each step to reach the globally optimal solution.

### Local Search Algorithm

Local Search starts with an arbitrary complete solution (e.g. random or greedy) and then iteratively makes small modifications to the solution to improve the value.

- Advantages
    - We can run as many iterations as we can afford to, which increases the likelihood of getting the the optimal value
    - At any point we can stop the iterations and can get a complete solution (which may not be the optimal one)

- Disadvantages
    - If the starting solution turns out to be an especially bad choice, it may need a large number of iterations to get to a good solution
