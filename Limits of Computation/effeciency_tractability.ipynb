## **Effeciency**
- Efficient: If the algorithm has a polynomial time complexity [O(n^k)] or lower
- Inefficient: If the algorithm has exponential [O(k^n)] or factorial [O(n!)]

## **Tractability/Solvability**
- Tractable/Solvable: A problem is considered tractable or solvable if there is a known solution with polynomial complexity or lower
- Otherwise, the problem is considered intractable
