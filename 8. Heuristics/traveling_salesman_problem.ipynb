# **Traveling Salesman Problem**

When given a fully connected weighted graphs, find the shortest cycle that visits every node exactly once and then returns to the start.

### **Exhaustive Approach [O(n!)]**

In an exhaustive approach, for a map with n cities the number of possible cycles would be (n-1)!/2. This is because each possible tour is a cycle, meaning that the number of cyclical tours would be n!/n=(n-1)!. Assuming it is an undirected graph between any two cities, both directions have equal distance.

### **Random Search [O(kn)]**
Consider some number k of randomly chosen tours, the complexity would be O(kn). Depending on chance, the results may get close or far from the optimal value.

### **Greedy Approach [O(n^2)]**
Construct solution step by step by optimizing local decisions. However, in the intermediate steps we do not have a complete solution. It has the advantage of being simple to implement and easy to run.

### **Local Search Algorithm [O(kn^2)]**

Starts by defining a neighbourhood with a k-opt. In an edge exhange neighbourhood, we delete _k_ edges in the current tour and then add _k_ edges which form a new feasible tour.

As each iteration has a worst-case run time of O(n^2), the total k-opt would be O(kn^2).
