# **Heaps**

In order for a binary tree to be considered a heap, it must fulfil 2 conditions:

1) **It must be a complete binary tree**: every non-leaf level of the tree must have the maximum number of nodes AND all leaf nodes must as leftmost as possible
2) **Heap-Order Property**: the value in every node must be greater than its parent (except the root node)



What is the worst-case time complexity of inserting 𝑛 keys into an empty heap if the keys are inserted one by one?

O(nlogn)

## Build heap from an unsorted array

when you know how many leaf nodes there'll be (l), take the last l values in the array and put them at the lowest level. Then take the next l/2 and insert them into the level above, swapping as necessary. Continue until all levels of the heap are filled. It has a worst case complexity of O(n).
