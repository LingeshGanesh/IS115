# **Graphs**

## **Directed Graphs**

A directed graph is a graph in which the edges have a direction associated with them, indicating a one-way relationship between the vertices. In directed graphs, there is a concept of "parent" and "child" vertexes with from=parent and to=child

Because of this, they have two different degree measures: indegree and outdegree

1) **indegeree:** The number of incoming edges to a vertex
2) **outdegree:** The number of outgoing edges from a vertex

Paths in a directed graph follow the direction of the edges and can be used to analyze the reachability between vertices. There are two possible types of path in graphs: paths and simple paths

- **Path:** A sequence of vertices connected by edges
- **Simple Path:** A path with a unique set of vertices

## Cycles

A directed graph may contain cycles, which are paths that start and end at the same vertex and contain at least one edge. This is important as certain traversal algorithms may fail if there are cycles present in the graph.

## **Undirected Graphs**

An undirected graph is one in which the edges are bidirectional in nature, with a two-way relationship between the vertices. Hence, there is no concept of a "parent" or "child" vertx. There is also only one type of degree measure as the degree of each vertex is simply the total number of edges connected to it.
