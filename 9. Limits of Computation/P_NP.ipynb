# **P vs NP**
#### Polynomial (P) vs Nondeterministic Polynomial (NP)

**Polynomial (P):** A problem is in the P if there is a known polynomial time algorithm to solve it

**Nondeterministic Polynomial (NP):** A problem is in NP if there is a known polynomial time algorithm to verify a given solution

An NP problem is considered NP-complete is all other NP problems can be reduced to it. If we can find a solution to an NP-complete problem, it implies that we can find a solution to all NP problem.

**Verifiable within Polynomial Time**
- [P]: Solvable in polynomial time, verifiable in polynomial time
- [NP]: No known polynomial time solution, verifiable in polynomial time
- [NP-complete]: No known polynomial time solution, verifiable in polynomial time, 
  every other NP problem can be reduced to it

**Not Verifiable within Polynomial Time**
- [NP-hard]: No known polynomial time solution, not necessarily verifiable 
  in polynomial time, every other NP problem can be reduced to it
