# **Hashmaps**

## **Hash Tables**

A hash table is a table where items are stored according to their keys using a hash function. Key values should be unique. A common hashing system is the Radix-26 or Base-26 numbering system. It uses letters in the alphabet A-Z as digits corresponding to 0-25 in base-10.

## Hash Function

The hash system converts a word into a number based on its letters. how many letters are used for the hash value may vary. In some cases when the hash value may be very large, we can use %n, where n is an arbitrary number of rows to make it smaller. The distribution of words tends to work betwen when n is a prime number.

## **Collisions**

As it is possible for two or more words to have the same hash value, collisions would occur in such cases. There are 2 main approaches to solving collisions: chaining (bucketing) and linear probing.

## Seperate Chaining (Buckets)

One approach is to store the values as a list of words (buckets) instead of storing just one value at each index. In this approach, each bucket is a secondary container or an unsorted list. The lookup would then be within the individual bucket after hashing. The worst case time complexity would be O(n) in the event that every key hashes to the same bucket.

## Linear Probing

The other apprach reuses empty spaces in the main table. This also has a worst case time complexity of O(n) as the worst case would be all hashing to the same spot so you would just get the same starting array.
