# **Uniform Cost Search (UCS)**

Uniform Cost Search is a pathfinding algorithm that expands the least cost node first, ensuring that the path to the goal node has the minimum cost. Unlike other search algorithms like Breadth-First Search (BFS), UCS takes into account the cost of each path, making it suitable for weighted graphs where each edge has a different cost.

UCS uses a similar approach to Dijkstras just that it uses an early stopping condition. While UCS stops once it has reached the goal node, Dijkstras computes the shortest paths to every node. Both are unable to handle negative edges.

### **Time complexity**
UCS has a time complexity big O notation of O((V+E)log V).

In the worst case, every node gets popped from the queue at most once. As there are V nodes we will do at most V pops. As each pop from the heap requires a reshuffle to find the new minimum, the time complexity would be O(V(log V)).

Similarly, in the worst case every edge in the graph gets looked at once, meaning that we will do E neighbour checks. As each check requires pushing onto the heap which would cost O(E(log V)).

Hence, the total cost would be O(V(log V)) + O(E(log V)) = O((V+E)log V).

### **Space Complexity**

When using the method of implementation shown in the slides, the space complexity has a big O of O(V). While the graph has an overall space complexity of O(V+E), when running UCS we only have an overhead of O(V) for the priority queue. The visited flag, cost value and previous pointers are all stored on each node object.

## Key Concepts:
1) **Priority Queue**: UCS uses a priority queue to store nodes. After initialization, the node with the lowest cumulative cost is expanded upon first. The idea is to explore the most promising paths first
2) **Path Cost**: The cost associated with getting to that node from the **start node**
3) **Exploration**: UCS explores nodes by expanding the least costly node first, continuing this process until the goal node is reached
