# **Topological Sort**

## **Directed Acyclic Graph (DAG)**

DAGs are directed graphs that do not contain cycles. They are used to show how the vertexes are related or depend on each other

## **Topo Sort (DFS)**

To determine the topological order of a graph, we perform DFS and add a node to a stack when it has been fully explored. Other nodes in the graph not connected to the starting node can be placed earlier in the topo order. Isolated nodes can be placed at any point in the topo order.
