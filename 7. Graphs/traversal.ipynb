# **Traversing a Graph**
https://www.youtube.com/watch?v=pcKY4hjDrxk

In order to understand traversal, there are 2 main key terms:

- Visiting a vertex: Going onto a particular vertex
- Exploration of a vertex: Visiting all adjacent vertices

Both approaches are O(n), where n is the number of vertices

## **Breadth-First Search (BFS)**

In BFS, we explore each vertex fully before moving on to the next one.

## Implementation

The list of vertices are stored in a FIFO queue, always expanding on the closest node.

## **Depth-First Search**

In DFS, we do not explore each vertex fully before moving on. Instead, we start exploring each new node only until the first vertex is found.

1) Look at the **FIRST** vertex connected to the starting vertex
2) Look at that vertex and look at its connected vertices
    - if there is only one vertex that has not been visited, visit it and check if it has any vertices
        - if it does, visit it and repeat step 2 on this new vertex
    - if there is more than one vertex that hasn't been visited, visit it and check if it has any vertices
        -  if it does, visit it and repeat step 2 on this new vertex
    - if all of its vertices have been visited
        - go back to the previous vertex and repeat step 2 until all its vertices have been visited

DFS usually uses a stack to keep track of who has been visited but not explored. The algorithm will know when it is traversed when the stack is empty.
